```python
DATA_ROOT = "./datasets/SCB5/SCB5-Handrise-Read-write-2024-9-17"

yaml_text = f"""path: {DATA_ROOT}
train: images/train
val: images/val
names:
  0: hand-raising
  1: read
  2: write
  3: bow the head
"""
with open("scb5.yaml","w", encoding="utf-8") as f:
    f.write(yaml_text)

!sed -n '1,120p' scb5.yaml
!ls -R "$DATA_ROOT" | head -n 50
```

```python
IMGSZ = 512
EPOCHS = 30

!yolo detect train \
  data=scb5.yaml model=yolo11n.pt \
  imgsz={IMGSZ} epochs={EPOCHS} batch=-1 device=0 \
  workers=2 cache=ram \
  mosaic=0.0 copy_paste=0.0 mixup=0.0 hsv_h=0.0 hsv_s=0.25 hsv_v=0.25 \
  patience=10 cos_lr=True
```
