```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
img = cv2.imread('/content/drive/MyDrive/Colab Notebooks/image/school.jpg')
net = cv2.dnn.readNet('/content/drive/MyDrive/Colab Notebooks/etc/la_muse.t7')
```

```python
img.shape
h, w, c = img.shape
```

```python
img = cv2.resize(img, dsize=(500, int(h / w * 500)))
print(img.shape)
```

```python
cv2_imshow(img)
```

```python
MEAN_VALUE = [103.939, 116.779, 123.68]
blob = cv2.dnn.blobFromImage(img, mean=MEAN_VALUE)
```

```python
blob.shape
```

```python
net.setInput(blob)
```

```python
output = net.forward()
```

```python
output.shape
```

```python
output = output.squeeze().transpose(1, 2, 0)
output.shape
```

```python
output += MEAN_VALUE
```

```python
output = np.clip(output, 0, 255)
output = output.astype('uint8')
```

```python
cv2_imshow(output)
```
