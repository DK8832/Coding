```python
img = cv2.imread('/content/drive/MyDrive/Colab Notebooks/image/amogus.jpg')
print(img.shape)
resize_img = cv2.resize(img, (300, 400))
print(img.shape)

print(img.shape)

resize_img = cv2.resize(img, (300, 400))
print(img.shape)

left_eye = (125, 110)
right_eye = (215, 110)
radius = 35

cv2.circle(resize_img, left_eye, radius, (0,0,0), 5)
cv2.circle(resize_img, right_eye, radius, (0,0,0), 5)

cv2.line(resize_img,
         (left_eye[0]+radius, left_eye[1]),
         (right_eye[0]-radius, right_eye[1]),
         (0,0,0), 5)
cv2.waitKey(0)
cv2_imshow(resize_img)
```
