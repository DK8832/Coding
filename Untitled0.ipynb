```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.preprocessing.image import img_to_array
from tensorflow.keras.models import load_model
```

```python
face_net = cv2.dnn.readNet("/content/drive/MyDrive/Colab Notebooks/2026.06.01/deploy.prototxt", "/content/drive/MyDrive/Colab Notebooks/2026.06.01/res10_300x300_ssd_iter_140000.caffemodel")
gender_net = cv2.dnn.readNet("/content/drive/MyDrive/Colab Notebooks/2026.06.01/deploy_gender.prototxt", "/content/drive/MyDrive/Colab Notebooks/2026.06.01/gender_net.caffemodel")
age_net = cv2.dnn.readNet("/content/drive/MyDrive/Colab Notebooks/2026.06.01/deploy_age.prototxt", "/content/drive/MyDrive/Colab Notebooks/2026.06.01/age_net.caffemodel")
```

```python
img = cv2.imread("/content/drive/MyDrive/Colab Notebooks/2026.06.01/peoples.jpg")
cv2_imshow(img)
```

```python
gender_list = ['Male', 'Female']
age_list = ['(0, 2)', '(4, 6)', '(8, 12)', '(15, 20)', '(25, 32)', '(38, 43)', '(48, 53)', '(65, 100)']
```

```python
h, w, c = img.shape
```

```python
blob = cv2.dnn.blobFromImage(img, 1.0, (300, 300), (104, 177, 123))
```

```python
face_net.setInput(blob)
```

```python
dets = face_net.forward()
```

```python
for i in range(dets.shape[2]):
  confidence = dets[0, 0, i, 2]
  if confidence < 0.6:
    continue

  x1 = int(dets[0, 0, i, 3] * w)
  y1 = int(dets[0, 0, i, 4] * h)
  x2 = int(dets[0, 0, i, 5] * w)
  y2 = int(dets[0, 0, i, 6] * h)




  face = img[y1:y2, x1:x2]
  blob = cv2.dnn.blobFromImage(face, 1.0, (227, 227), mean=(78.4263377603, 87.7689143744, 114.895847746))
  gender_net.setInput(blob)
  gender_index = gender_net.forward().squeeze().argmax()
  gender = gender_list[gender_index]
  if gender_index == 0:
    color = (255, 0, 0)
  else:
    color = (0, 0, 255)
  age_net.setInput(blob)
  age_index = age_net.forward().squeeze().argmax()
  age = age_list[age_index]
  cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
  cv2.putText(img, text='%s %s' % (gender, age), org=(x1, y1-15), fontFace=cv2.FONT_HERSHEY_COMPLEX, fontScale=0.6, color=color, thickness=2)
cv2_imshow(img)
```
